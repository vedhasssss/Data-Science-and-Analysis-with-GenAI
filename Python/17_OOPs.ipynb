```python
#class-> is a bluprint or template to create a object.It defines the behaviour of the the object. Defined using keyword "class"
class customers():
    def bankname():
        print("SBI")
    def creditlimit():
        print("creditLimit:",8000)
```

```python
#Object->An object is a specific instance of a class. 
customer_obj=customers #creating object

customer_obj.bankname()#calling using object
customer_obj.creditlimit()
```

```python
class employee:
    def emp_salary(salary):
        print("Employee salary :",salary)

    def emp_name(name):
        print("Employee Name:",name)

employee.emp_salary(30000)
employee.emp_name("Vedhas")
```

```python
class employee1:
    def __init__(self): # defining Constructor -> gets called automatically (no need to call)
        print("Infosys Technologies")

    def emp_salary(self,salary):
        print("Employee",salary)

    def emp_name(self,name):
        print("Employee Name:",name)

emp=employee1() #Constructor
emp.emp_name("Vedhas")
emp.emp_salary(45000)

```

```python
class Employee2:
    def __init__(self): #self creates a bridge between functions (so tha they are connected to each other)
        pass

    def tds(self,tax):
        self.tax=tax #so other function can access the data or variable in a class
        print("Tax:",tax)

    def emp_salary(self,salary):
        print("Employee Salary:",salary-self.tax) #access parameter of other function

    def emp_name(self,name):
        print("Employee Name:",name)

emp=Employee2()
emp.tds(1000)
emp.emp_name("Vedhas")
emp.emp_salary(45000)
```
