```python
#function with default parameter
def bill(price,gst=18):
    gst_amount=gst*price/100
    total_amount=price+gst_amount
    print("\n======bill========")
    print(f"Price={price} Rs")
    print(f"GST={gst}%")
    print(f"Gst ammount{gst_amount} Rs")
    print(f"Total amount={total_amount}")
```

```python
bill(1200)
bill(100,9)
```

```python
#Lambda Function -> Anonymous function
#Lambda Agrgument , Expression

add=lambda a , b : a+b
print(add(20,30))
```

```python
sub=lambda a , b : a-b
print(sub(20,30))
```

```python
division=lambda a , b : a/b
print(division(30,6))
```

```python
multiplication = lambda a,b:a*b
print(multiplication(20,10))
```

```python
#date and Time
import datetime
```

```python
# get current date time
current = datetime.datetime.now() #all
print(current)


# get current date 
today = datetime.date.today() #only date
print(today)

# get current time 
time = datetime.datetime.now().time() # only time
print(time)
```

```python
from datetime import date
# create a custom date
my_date = date(2026,11,4)
print(my_date)
print("year =  " ,my_date.year)
print("month =  " ,my_date.month)
print("DAY = ",my_date.day)
```

```python
current = datetime.datetime.now()
print(current)
print("YEAR = ",current.year)
print("MONTHS = ",current.month)
print("DAY =",current.day)
print("HOUR =",current.hour)
print("MINUTES = ",current.minute)
print("SECONDS = ",current.second)
```
