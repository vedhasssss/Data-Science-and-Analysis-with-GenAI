```python
#Zip
names=['jay','Roy','Toy']
marks=[34,40,36]
a=zip(names,marks)
print(list(a))
```

```python
#zip using 3 list
names=['jay','Roy','Toy']
marks=[34,40,36]
subject=['Python','Java','C++']

a=zip(names,marks,subject)

print(list(a))
```

```python
#dictionary in zip
keys=['Name','Age','City']
values=['Vedhas',21,'Mumbai']

a=dict(zip(keys,values))
print(a)
```

```python
#zip_longest()
from itertools import zip_longest
names=['jay','Roy','Toy','Amit']
marks=[34,40,36]
a=zip_longest(names,marks,fillvalue='-')
print(list(a))
```

```python
#Enumerate() -> returns index of elements with pair with it
names=["joy","Toy","Roy","chaman","Scout"]
result=enumerate(names)
print(list(result))
```

```python
#Revision of some topics
#lambda
discount_fare=lambda a:a-200 #200 rs discount
print("Discounted fare=",discount_fare(2500))
```

```python
#map
ticket_fares=[2000,7000,3400]
#increasing fares by 100 rs
inc_fare=lambda a:a+100
result=map(inc_fare,ticket_fares)
print(list(result))
```

```python
#Filters
status=['Confirmed','Waiting',"RAC","Confirmed"]
x=lambda a:a =="Confirmed"
result=filter(x,status)
print(list(result))
```

```python
passeneger=["Yung","Scout","Love","Gullu"]
seats=["B1-21","B1-22","B1-23","B1-24"]
x=zip(passeneger,seats)
print(list(x))
```

```python
z=enumerate(passeneger)
print(list(z))
```

```python
#process data using filters() and maps()
ticket_fares=[2500,3000,4000,1500,1000,800]
z=lambda a:a>=1500
valid_fares=list(filter(z,ticket_fares))
print(valid_fares)
```

```python
#Apply fare update
#x=lambda a:a +100
final_fare=map(lambda a:a+100,valid_fares)
print(list(final_fare))
```
