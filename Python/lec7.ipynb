```python
#list
#square brackets are used for list
#Used to store multiple values in a variable
list_1=[1,2,3,4,5]
print(list_1)
list_2=["a","b","c"]
print(list_2)
list_3=[1,2,3,4,['a','b','c']] #list ke andar list
print(list_3)

print(type(list_3))
```

```python
#in python indexing starts from 0 , like 0,1,2,3,4,..... so on
data=[1,2,34,55,6,7,4,3,222,565,90]
print(data[2]) #accessing a value from a list
print(data[-2]) #reverse accessing
```

```python
#slicing
num=[12,34,43,24,56,76,55,87]
print(num[2:5]) #accessing in between values from list(excludes last mentioned index)
print(num[3:]) #accessingg after an index
print(num[:6]) #accessing before an index
```

```python
#updating data
data=[23,35,67,89,56]
data[3]=100
print(data)
```

```python
#List Methods

#1. append() -> used to insert an value in last of the list
data=[22,33,45,67,87,92,34]
data.append(400)
print(data)
```

```python
#2. extend() -> inserting multiple values at end of the list
data=[22,33,45,67,87,92,34]
data.extend([20,30,40])
print(data)
```

```python
#3. insert() -> used to inesrt a value in between of the list
data=[22,33,45,67,87,92,34]
data.insert(4,1)
print(data)
```

```python
#4. pop() <-Removes an element by index
data=[22,33,45,67,87,92,34]
data.pop(3)
print(data)
```

```python
#5. remove() <-Removes a mentioned element
data=[22,33,45,67,87,92,34]
data.remove(45)
print(data)
```

```python
#6.index -> returns the index of an element
data=[22,33,45,67,87,92,34]
print(data.index(67))
```

```python
#7.count() -> returns count of an element(duplicate)
data=[22,33,45,67,87,92,33,34]
print(data.count(33))
```

```python
#8.sort -> descending order me sort karta hai
data=[22,33,45,67,87,92,34]
data.sort()
print(data)
```

```python
#9. reverse -> reverses the list
data=[22,33,45,67,87,92,34]
data.reverse()
print(data)
```

```python
#10. copy -> copies the whole list
data=[22,33,45,67,87,92,34]
data1=data.copy()
print(data1)
```

```python
#11. clear -> clears the list
data=[22,33,45,67,87,92,34]
data.clear()
print(data)
```

```python
#len, sun , min , max
data=[22,33,45,67,87,92,34]
print(len(data))
print(sum(data))
print(min(data))
print(max(data))
```
