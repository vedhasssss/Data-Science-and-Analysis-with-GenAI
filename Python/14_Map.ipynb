```python
#Lambda 
max=lambda a,b:a if a>b else b 
max(34,33)
```

```python
#lambda with string
get_len=lambda a:len(a)
len("Hello")
```

```python
#lambda with list
num=[23,33,1,2,3,4,5,9]

sq=lambda a: a*a
result=[sq(a) for a in num]
print(result)
```

```python
#MAP
#map(function,iterable)
#iterable-> list,tuple

#without map
numbers=[23,54,56,21,34,57,98]
result=[]
for i in numbers:
    result.append(i*i)

print(result)
```

```python
#with map
numbers=[23,54,56,21,34,57,98]

def square(num):
    return num*num

result=map(square,numbers)

print(list(result))
```

```python
#using map and lambda
numbers=[23,54,56,21,34,57,98]
result=map(lambda a:a*a,numbers)
print(list(result)) # converting result into list
```
