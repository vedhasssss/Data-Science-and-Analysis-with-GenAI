```python
#Strings Operations
''' 
1.upper()
2.lower()
3.partition
4.replace()
5.find()
6.rstrip()
7.split()
8.startswith()
9.isnumeric()
10.index()
'''
```

```python
#upper() -> coverts string into uppercase
data="Code with Vedhas"
print(data.upper())
```

```python
#lower() -> converts string into lowercase
print(data.lower())
```

```python
#partition -> creates the partition
print(data.partition("w"))
```

```python
#replace -> replaces the string
print(data.replace("Vedhas","Python"))
```

```python
#index -> returns the index of the string
print(data.index("w"))
```

```python
#find() -> same as index
print(data.find("C"))
```

```python
#rstrip -> clears the blank space
data_1="Vedhas   "
print(data_1,10)
print(data_1.strip(),10)
```

```python
#split() ->splits the string
data_2="a b c d e f g h i j k l m n o p q r s t u v w x y z"
print(data_2.split(" "))
```

```python
#startswith() -> check whether string starts with certain string or char by returning boolean value
print(data.startswith("C"))
print(data.startswith("w"))
```

```python
#isnumeric() -> checks if the string is number or not
num="67"
print(num.isnumeric())
```

```python
#format()
Name="Vedhas"
Age=21
height=5.11

#normally
print("My name is",Name,"age is",Age,"Height is",height)

#using format()
print("My name is {0} Age is {1} Height is {2}".format(Name,Age,height))

#using f
print(f"My name is {Name} Age is {Age} Height is {height}")
```
