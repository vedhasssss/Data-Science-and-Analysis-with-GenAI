```python
#Tuples -> not mutable , duplicates are allowed
a=56
print(type(a))

#tuple
a=56,
print(type(a))

a=(1,2,3,4,5,6,77,77,88,88)
print(type(a))

b=23,44,55,66,77,77,88
print(type(b))
```

```python
#Concatination
tuple1=(54,65,87)
tuple2=(4,5,6)
result=tuple1+tuple2
print(result)
```

```python
#count()
a=(1,2,3,4,5,6,77,77,88,88)
print(a.count(77))
```
