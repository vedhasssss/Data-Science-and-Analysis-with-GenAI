```python
# for loop
# range(end_point)
for i in range(10):
    print(i)
```

```python
#range(start_point,end_point)
for i in range(10,20):
    print(i)
```

```python
#range(start_point,end_point,diffrence/gap)
for i in range(0,50,5):
    print(i)
```

```python
#for horizontal
for i in range(1,10):
    print(i,end=" | ")
```

```python
#reversing (-1)
for i in range(10,0,-1):
    print(i)
```

```python
# reversing part-2 --> range(-start_point,end_point) and print  (-i)
for i in range(-10,0):
    print(-i)
```

```python
#while loop
x=0
while x<=100:
    print(x)
    x=x+10
```
