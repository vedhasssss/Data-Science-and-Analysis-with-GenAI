```python
#--------------------SET----------------------
'''
Every item is uique
order does not matter
Its mutable
Automatically removes duplicates
faster than list
'''
```

```python
#empty set
data=set()
print(data)
print(type(data))
```

```python
#adding in set
data={12,34,54,32,19}
print(data)
data.add(100)
print(data)
```

```python
#adding multiple data
data.update([200,300,400])
print(data)
```

```python
#remove -> if it finds the item then removes or else throws error and terminates the program(dangerous option)
data.remove(54)
print(data)
```

```python
#discard -> Removes item and does not throw error when item not found (safer option)
data.discard(54)
print(data)
```

```python
#pop() -> removes random item (dangerous)
data.pop()
print(data)
```

```python
#clear() -> removes all items
data.clear()
print(data)
```
