```python
#dictionary -> stores data in key value pair
my_dict={
    "emp_id":33,
    "Name":"Vedhas",
    "Age":20,
    "Organisation":"Google"
}
print(my_dict)
print(type(my_dict)) #check data type

print(my_dict["Age"]) #accessing key's value
```

```python
#Modifying value 
my_dict["Age"]=21
print(my_dict["Age"])
```

```python
#Adding new key:value
my_dict["city"]="Mumbai"
print(my_dict)
```

```python
#Checking key existence
if "Marks" in my_dict:
    print("Marks found")
else:
    print("Marks not found")

if "city" in my_dict:
    print("City found")
else:
    print("City not found")
```

```python
#keys -> returns a list of all keys in the dictionary
print(my_dict.keys())

#values ->Returns a list of all values from the dictionary
print(my_dict.values())

#items ->Returns a list of all key value pairs as tuples
print(my_dict.items())

#get() ->Returns a value of an key
print(my_dict.get("Name"))
```

```python
#pop -> Remove an key value pair from the dictionary
my_dict.pop("city")
print(my_dict)
```

```python
#using loop
for i in my_dict:
    print(i,"=",my_dict[i]) #here i returns key and my_dict[i] returns value
```

```python
#Strings
data="Apple"
print(data)
print(type(data))
print(data[2])
print(data[1:4])
print(data[3:])
print(data[:3])
```

```python
#using for loop
for i in data:
    print(i)
```
