```python

#Arithmetic Operator
x=25
y=3
print(x+y)#Addition
print(x-y)#Subtraction
print(x*y)#Multiplication
print(x/y)#Division
print(x//y)#no decimal values division
print(x%y)#Modulas
print(2**3,"\n \n ")#Power

```

```python
#Comparison Operators
print(23==23)#is equal to
print(23!=23)#is not equal to
print(25>23)#greater than
print(23<24)#less than
print(23>=21)#Greater than equal to
print(23>=23)
print(23<=21)#Less than equal to
```

```python
#Logical Operator
print(45<50 and 67!=67)#and 

print(45<90 or 67!=67)#or

print(not 34==34) #not
```

```python
#Membership operator
name="Vedhas"
name1="Ved"
print("a" in name) #in
print("a" not in name)#not in
print(name is name1)#is
print(name is not name1)#is not
```
