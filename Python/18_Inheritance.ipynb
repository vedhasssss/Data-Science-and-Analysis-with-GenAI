```python
#Example
class bank:
    def bank_info():
        print("Axis Private Limited")
        print("Borivali west,Gorai-2\nMumbai-400091")
```

```python
class customer(bank):
    def customer_info():
        print("Customer Name:- Vedhas Shinde")
```

```python
obj=customer
obj.bank_info()
obj.customer_info()
```

```python
#inheritance with constructor (__init__)
class parent1:
    def __init__(self):
        print("Parent Constructor")

class child1(parent1):
    def __init__(self):
        super().__init__() # This is used to stop overlapping of contructor or method of parent class
        print("Child Constructor")
```

```python
obj1=child1()
```

```python
#Multiple Inheritance
class Father:
    def father_info(self):
        print("Father Name is Virat")

class Mom:
    def mom_info(self):
        print("Mother Name is Anushka")

class Child(Father,Mom):
    def child_info(self):
        print("Child Name is Vamika")
```

```python
obj_child=Child() # created an Object
obj_child.father_info()
obj_child.mom_info()
obj_child.child_info()
```

```python
# Multilevel Inheritance-> A child class inherits properties from a parent class whichh chid class of another parent class

class Grandfather:
    def grandfather_info(self):
        print("Grandfather Name is Prem")

class Father(Grandfather):
    def father_info(self):
        print("Father Name is Virat")

class Child(Father):
    def child_info(self):
        print("Child Name is Vamika")
```

```python
# Object Creation
c = Child()

c.grandfather_info()
c.father_info()
c.child_info()
```

```python
#Hierarchical Inheritance -> Multiple child class inherits properties of 1 parent class
class Parent11:
    def display1(self):
        print("Property Money:",3006)

class Child11(Parent11):
    def display2(self):
        print("Property got:",1001)

class Child12(Parent11):
    def display3(self):
        print("Property got:",1002)

class Child13(Parent11):
    def display4(self):
        print("Property got:",1003)

#Object Creations (3 object created)
c1=Child11()
c2=Child12()
c3=Child13()

c1.display1()
c1.display2()
c2.display3()
c3.display4()
```

```python
#hybrid Inheritance: any class can inherit properties from any parent class 
# Hierarchical Inheritance

class A:
    def class_a(self):
        print("This is Class A")

class B(A):
    def class_b(self):
        print("This is Class B")

class C(A):
    def class_c(self):
        print("This is Class C")
```

```python
# Object Creation
b = B()
c = C()

b.class_a()
b.class_b()

c.class_a()
c.class_c()
```
